```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import classification_report, confusion_matrix, accuracy_score
```

```python
df=pd.read_csv("cleaned_creditdata.csv")
```

```python
df
```

```python
df.duplicated().sum()
```

```python
remove_dup= df.drop_duplicates()
```

```python
remove_dup.count()
```

```python
df
```

```python
df["loan_int_rate"] = df["loan_int_rate"].fillna(df["loan_int_rate"].median())
df["person_emp_length"] = df["person_emp_length"].fillna(df["person_emp_length"].median())
```

```python
df.isnull().sum()
```

```python
X=df[["person_age","person_income","person_emp_length","loan_amnt","loan_int_rate","cb_person_cred_hist_length"]]
Y=df["loan_status"]
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,Y_train,Y_test=train_test_split(X,Y,test_size=0.2,random_state=0)
```

```python
from sklearn.preprocessing import StandardScaler
scale=StandardScaler()
X_train_transformed=pd.DataFrame(scale.fit_transform(X_train),columns=X_train.columns,index=X_train.index)
X_test_transformed=pd.DataFrame(scale.fit_transform(X_test),columns=X_test.columns,index=X_test.index)
print(X_train_transformed)
```

```python
X_train_transformed.info()
```

```python
X_test_transformed.info()
```

```python
from sklearn.tree import DecisionTreeClassifier
classifier=DecisionTreeClassifier()
classifier.fit(X_train_transformed,Y_train)
```

```python
y_test_pred=classifier.predict(X_test_transformed)
```

```python
DT_accuracy=(accuracy_score(Y_test,y_test_pred))
print(classification_report(Y_test,y_test_pred))
DT_accuracy
```

```python
from sklearn.neighbors import KNeighborsClassifier
KNclassifier=KNeighborsClassifier()
KNclassifier.fit(X_train_transformed,Y_train)
```

```python
y_test_pred=KNclassifier.predict(X_test_transformed)
kn_accuracy=(accuracy_score(Y_test,y_test_pred))
print(classification_report(Y_test,y_test_pred))
kn_accuracy
```

```python
from sklearn.ensemble import RandomForestClassifier
classifier=RandomForestClassifier()
classifier.fit(X_train_transformed,Y_train)
```

```python
y_test_pred=classifier.predict(X_test_transformed)
RF_accuracy=(accuracy_score(Y_test,y_test_pred))
print(classification_report(Y_test,y_test_pred))
RF_accuracy
```

```python
from sklearn.svm import SVC
SVclassifier=SVC()
SVclassifier.fit(X_train_transformed,Y_train)
```

```python
y_test_pred=SVclassifier.predict(X_test_transformed)
SVC_accuracy=(accuracy_score(Y_test,y_test_pred))
print(classification_report(Y_test,y_test_pred))
SVC_accuracy
```

```python
from sklearn.ensemble import BaggingClassifier
BAclassifier=BaggingClassifier()
BAclassifier.fit(X_train_transformed,Y_train)
```

```python
y_test_pred=BAclassifier.predict(X_test_transformed)
BA_accuracy=(accuracy_score(Y_test,y_test_pred))
print(classification_report(Y_test,y_test_pred))
BA_accuracy
```

```python
from sklearn.linear_model import LogisticRegression
classifier=LogisticRegression()
classifier.fit(X_train_transformed,Y_train)
```

```python
y_test_pred=classifier.predict(X_test_transformed)
log_accuracy=(accuracy_score(Y_test,y_test_pred))
print(classification_report(Y_test,y_test_pred))
log_accuracy
```

```python
from xgboost import XGBClassifier
xgb_classifier = XGBClassifier(n_estimators=100, max_depth=6, learning_rate=0.1, random_state=0, use_label_encoder=False, eval_metric='logloss')
xgb_classifier.fit(X_train_transformed, Y_train)

y_test_pred_xgb = xgb_classifier.predict(X_test_transformed)
xgb_accuracy = accuracy_score(Y_test, y_test_pred_xgb)
print(classification_report(Y_test, y_test_pred_xgb))
xgb_accuracy
```

```python
evaluation = pd.DataFrame({'Algorithms':['Logistic Regression','Decision Tree','Random Forest','SVC','KNN','Bagging','XGBoost'],"Accuracy":[log_accuracy,DT_accuracy,RF_accuracy,SVC_accuracy,kn_accuracy,BA_accuracy,xgb_accuracy]})
```
